````python
# Analysis of Study Time and Student Performance

**Course:** B.Sc. (Hons.) Data Science - Semester I
**Name:** Ritesh
**Roll No.:** 2601840023
**Domain:** Education
**Institution:** K.R. Mangalam University

---

## 1. Executive Summary / Abstract
This project examines whether the time a student spends studying each week is directly related to their performance in final test scores. Using a clean student performance dataset, the analysis follows five core phases: defining the problem, understanding dataset attributes, data preparation & cleaning, statistical EDA, and predictive modeling. Results show a statistically detectable positive relationship between study time and final test scores, alongside the significant influence of prior academic performance.

---

## 2. Problem Statement
The amount of time students spend studying may be related to their performance in assessments. This project analyzes student data containing study-time information and final test scores. It calculates summary statistics, compares test scores across study-time ranges, and uses appropriate visualisations and correlation analysis to examine whether higher study time is associated with higher scores. The report also acknowledges other factors that may influence test performance and are not represented in the dataset.

---

## 3. Dataset Description
* **studytime:** Weekly study hours logged by the student (Numeric).
* **final_score:** Final exam/assessment score achieved (Numeric, target outcome).
* **previous_scores:** Past assessment scores (Numeric).
* **sleep_hours:** Average daily sleep duration (Numeric).

---

## 4. Methodology
1. **Data Cleaning:** Verified zero null values, removed duplicate rows, and checked value range consistency.
2. **Exploratory Data Analysis (EDA):** Generated distribution histograms, correlation heatmaps, and regression trend scatter plots.
3. **Statistical & ML Modeling:** Evaluated Spearman rank correlation ($\rho$) and trained a Linear Regression model with an 80/20 train-test split.

---

## 5. Key Findings & Limitations
* **Key Finding:** Weekly study time exhibits a positive correlation with test performance. Increased study duration consistently correlates with improved final assessment scores.
* **Limitations:** The dataset does not measure qualitative metrics such as study environment quality, student motivation, attendance quality, or mental well-being, which also strongly impact final test results.

---

## 6. Project File Structure
```text
├── Ritesh_2601840023_Capstone.ipynb   # Executed Jupyter/Colab notebook
├── Student_Performance.csv            # Cleaned input dataset
└── README.md                          # Project documentation summary
````
